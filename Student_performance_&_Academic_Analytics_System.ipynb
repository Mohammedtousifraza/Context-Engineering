Student Dataset

```python
students = [
    {
        "id": 1,
        "name": "Mohammed",
        "python": 85,
        "dbms": 78,
        "ai": 91,
        "mathematics": 88
    },
    {
        "id": 2,
        "name": "Aisha",
        "python": 72,
        "dbms": 81,
        "ai": 76,
        "mathematics": 79
    },
    {
        "id": 3,
        "name": "Rahul",
        "python": 91,
        "dbms": 89,
        "ai": 94,
        "mathematics": 90
    },
    {
        "id": 4,
        "name": "Sneha",
        "python": 65,
        "dbms": 70,
        "ai": 68,
        "mathematics": 74
    },
    {
        "id": 5,
        "name": "Arjun",
        "python": 55,
        "dbms": 61,
        "ai": 58,
        "mathematics": 63
    }
]

print("Number of students:", len(students))
```

Calculate Student Performance

```python
def calculate_performance(student):
    total = (
        student["python"]
        + student["dbms"]
        + student["ai"]
        + student["mathematics"]
    )

    percentage = total / 4

    if percentage >= 90:
        grade = "A+"
    elif percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"

    if percentage >= 40:
        result = "PASS"
    else:
        result = "FAIL"

    student["total"] = total
    student["percentage"] = percentage
    student["grade"] = grade
    student["result"] = result

    return student
```

Process All Students

```python
for student in students:
    calculate_performance(student)

print("Student performance calculated successfully.")
```

Display Student Results

```python
print("\n" + "=" * 80)
print("STUDENT PERFORMANCE REPORT")
print("=" * 80)

for student in students:
    print(f"""
ID          : {student["id"]}
Name        : {student["name"]}
Python      : {student["python"]}
DBMS        : {student["dbms"]}
AI          : {student["ai"]}
Mathematics : {student["mathematics"]}
Total       : {student["total"]}/400
Percentage  : {student["percentage"]:.2f}%
Grade       : {student["grade"]}
Result      : {student["result"]}
{"-" * 80}
""")
```

Find Top Student

```python
def find_top_student(students):
    top_student = students[0]

    for student in students:
        if student["percentage"] > top_student["percentage"]:
            top_student = student

    return top_student


top_student = find_top_student(students)

print("TOP PERFORMER")
print("=" * 40)
print("Name       :", top_student["name"])
print("Percentage :", f'{top_student["percentage"]:.2f}%')
print("Grade      :", top_student["grade"])
```

Class Statistics

```python
def calculate_class_statistics(students):
    total_percentage = 0
    passed = 0
    failed = 0

    for student in students:
        total_percentage += student["percentage"]

        if student["result"] == "PASS":
            passed += 1
        else:
            failed += 1

    class_average = total_percentage / len(students)
    pass_percentage = (passed / len(students)) * 100

    return class_average, passed, failed, pass_percentage


class_average, passed, failed, pass_percentage = calculate_class_statistics(students)

print("=" * 50)
print("CLASS STATISTICS")
print("=" * 50)
print("Total Students :", len(students))
print("Class Average  :", f"{class_average:.2f}%")
print("Passed         :", passed)
print("Failed         :", failed)
print("Pass Percentage:", f"{pass_percentage:.2f}%")
```

Subject Analysis

```python
def calculate_subject_average(students, subject):
    total = 0

    for student in students:
        total += student[subject]

    return total / len(students)


subjects = ["python", "dbms", "ai", "mathematics"]

print("=" * 50)
print("SUBJECT-WISE ANALYSIS")
print("=" * 50)

for subject in subjects:
    average = calculate_subject_average(students, subject)
    print(f"{subject.title():12} : {average:.2f}%")
```

Highest Performing Subject

```python
subject_averages = {}

for subject in subjects:
    subject_averages[subject] = calculate_subject_average(students, subject)

best_subject = max(subject_averages, key=subject_averages.get)

print("\nBEST PERFORMING SUBJECT")
print("=" * 40)
print("Subject :", best_subject.title())
print("Average :", f"{subject_averages[best_subject]:.2f}%")
```

Search Student

```python
def search_student(students, student_name):
    for student in students:
        if student["name"].lower() == student_name.lower():
            return student

    return None


search_name = input("Enter student name to search: ")

student = search_student(students, search_name)

if student:
    print("\nStudent Found")
    print("=" * 40)
    print("Name       :", student["name"])
    print("Percentage :", f'{student["percentage"]:.2f}%')
    print("Grade      :", student["grade"])
    print("Result     :", student["result"])
else:
    print("\nStudent not found.")
```

Add a New Student

```python
def get_valid_mark(subject):
    while True:
        try:
            mark = float(input(f"Enter {subject} marks (0-100): "))

            if 0 <= mark <= 100:
                return mark
            else:
                print("Marks must be between 0 and 100.")

        except ValueError:
            print("Please enter a valid number.")


print("=" * 50)
print("ADD NEW STUDENT")
print("=" * 50)

name = input("Enter student name: ")

new_student = {
    "id": len(students) + 1,
    "name": name,
    "python": get_valid_mark("Python"),
    "dbms": get_valid_mark("DBMS"),
    "ai": get_valid_mark("AI"),
    "mathematics": get_valid_mark("Mathematics")
}

calculate_performance(new_student)
students.append(new_student)

print("\nStudent added successfully!")
print("Name       :", new_student["name"])
print("Percentage :", f'{new_student["percentage"]:.2f}%')
print("Grade      :", new_student["grade"])
print("Result     :", new_student["result"])
```

API Integration


```python
import requests

api_url = "https://jsonplaceholder.typicode.com/users/1"

try:
    response = requests.get(api_url, timeout=10)

    print("HTTP Status Code:", response.status_code)

    if response.status_code == 200:
        user_data = response.json()

        print("\nAPI RESPONSE")
        print("=" * 40)
        print("Name    :", user_data["name"])
        print("Username:", user_data["username"])
        print("Email   :", user_data["email"])
        print("City    :", user_data["address"]["city"])
    else:
        print("API request failed.")

except requests.exceptions.RequestException as error:
    print("Error connecting to API:", error)
```

Final Report

```python
def generate_report(students):
    class_average, passed, failed, pass_percentage = calculate_class_statistics(students)
    top_student = find_top_student(students)

    print("\n")
    print("=" * 70)
    print("              FINAL ACADEMIC ANALYTICS REPORT")
    print("=" * 70)

    print("Total Students       :", len(students))
    print("Class Average        :", f"{class_average:.2f}%")
    print("Students Passed      :", passed)
    print("Students Failed      :", failed)
    print("Pass Percentage      :", f"{pass_percentage:.2f}%")
    print("Top Performer        :", top_student["name"])
    print("Top Performer Score  :", f'{top_student["percentage"]:.2f}%')

    print("\nSubject Averages")
    print("-" * 40)

    for subject, average in subject_averages.items():
        print(f"{subject.title():15}: {average:.2f}%")

    print("=" * 70)


generate_report(students)
```
